```python


import pandas as pd
import numpy as np

# Step 1: Create a sample dataset
data = {
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Ravi', 'Meena'],
    'Age': [22, np.nan, 25, 23, np.nan, 24],
    'Salary': [25000, 30000, np.nan, 28000, 32000, np.nan],
    'Department': ['IT', 'HR', 'IT', np.nan, 'Finance', 'HR'],
    'Experience': [1, 3, 2, np.nan, 5, 4]
}

df = pd.DataFrame(data)

# Step 2: Display the dataset
print("===== ORIGINAL DATASET =====")
display(df)

# Step 3: Check for missing values
print("\n===== MISSING VALUE SUMMARY =====")
print(df.isnull().sum())

# Step 4: Calculate missing values in each column
missing_counts = df.isnull().sum()

print("\n===== COLUMN-WISE MISSING VALUE COUNTS =====")
print(missing_counts)

# Step 5: Calculate missing value percentages
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\n===== MISSING VALUE PERCENTAGES =====")
print(missing_percentage.round(2))

# Step 6: Identify columns requiring treatment
print("\n===== COLUMNS REQUIRING TREATMENT =====")

columns_with_missing = missing_counts[missing_counts > 0]

if not columns_with_missing.empty:
    print(columns_with_missing)
else:
    print("No missing values found.")

# Step 7: Generate the final analysis report
summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage (%)': missing_percentage.round(2)
})

print("\n===== FINAL MISSING VALUE ANALYSIS REPORT =====")
display(summary)

# Step 8: Display total missing values
print("\nTotal Missing Values:", int(df.isnull().sum().sum()))

print("\nTask Completed Successfully!")
```
